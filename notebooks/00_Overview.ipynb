# DSPM + Zero Trust

This repo is **public-safe** and **complete**:
- Uses **synthetic input files** in `data/`
- Demonstrates DSPM lifecycle: **Discover → Classify → Protect → Monitor → Destroy (planned/HITL)**
- Translates DSPM outputs into **Zero Trust policies** (identity, posture, conditional access)

> ⚠️ Destroy is modeled as **eligibility + approvals + scheduling** only. No destructive actions.


## Inputs
- `data/synthetic_assets.csv`
- `data/synthetic_access_events.csv`
- `data/schema.json`

## Run order
Run notebooks `01` → `07` in order.
